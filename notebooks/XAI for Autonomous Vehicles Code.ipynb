```python

def detect_objects(image, model, conf_threshold=0.3):
    """
    Run object detection on an image
    
    Args:
        image: numpy array (H, W, 3) - BGR format
        model: YOLOv5 model
        conf_threshold: confidence threshold
    
    Returns:
        results: detection results
        inference_time: time taken in milliseconds
    """
    start_time = time.time()
    
    # Convert BGR to RGB for YOLO
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    
    # Run inference
    model.conf = conf_threshold  # Set confidence threshold
    results = model(image_rgb)
    
    inference_time = (time.time() - start_time) * 1000  # Convert to ms
    
    return results, inference_time


def parse_detections(results, driving_classes_only=True):
    """
    Parse YOLOv5 detection results
    
    Returns:
        detections: list of dicts with detection info
    """
    detections = []
    
    # Use pandas format - most reliable
    pred_df = results.pandas().xyxy[0]
    
    for idx, row in pred_df.iterrows():
        class_id = int(row['class'])
        
        # Filter for driving-relevant classes if requested
        if driving_classes_only and class_id not in DRIVING_CLASSES:
            continue
        
        # Extract coordinates (already in correct format)
        x1, y1, x2, y2 = row['xmin'], row['ymin'], row['xmax'], row['ymax']
        
        det = {
            'class_id': class_id,
            'class_name': row['name'],
            'confidence': float(row['confidence']),
            'bbox': np.array([x1, y1, x2, y2], dtype=np.float32),
            'center_x': (x1 + x2) / 2,
            'center_y': (y1 + y2) / 2,
            'width': x2 - x1,
            'height': y2 - y1
        }
        
        detections.append(det)
    
    return detections


# Test on latest captured frame
if latest_image is not None:
    print("Running detection on latest frame...")
    print("-" * 50)
    
    # Test with YOLOv5-Small
    results, inf_time = detect_objects(latest_image, models['yolov5s'])
    detections = parse_detections(results)
    
    print(f"Inference time: {inf_time:.2f}ms")
    print(f"Objects detected: {len(detections)}")
    print()
    
    # Display detections
    for det in detections:
        print(f"  {det['class_name']:15s} | Conf: {det['confidence']:.2f} | "
              f"Position: ({det['center_x']:.0f}, {det['center_y']:.0f})")
    
    print("-" * 50)
else:
    print("⚠️ No image available. Make sure vehicle and camera are running.")
```

```python
# ========================================
# CELL: Define Grad-CAM for YOLOv5
# ========================================

class YOLOv5GradCAM:
    """
    Simplified Grad-CAM implementation for YOLOv5 models
    """
    
    def __init__(self, model, target_layer):
        """
        Args:
            model: YOLOv5 model (AutoShape wrapper)
            target_layer: Layer to compute gradients
        """
        self.model = model
        self.target_layer = target_layer
        self.gradients = None
        self.activations = None
        
        # Register hooks
        self.hook_handles = []
        h1 = self.target_layer.register_forward_hook(self.save_activation)
        h2 = self.target_layer.register_full_backward_hook(self.save_gradient)
        self.hook_handles.extend([h1, h2])
    
    def save_activation(self, module, input, output):
        """Save forward pass activations"""
        self.activations = output.detach()
    
    def save_gradient(self, module, grad_input, grad_output):
        """Save backward pass gradients"""
        self.gradients = grad_output[0].detach()
    
    def generate_cam(self, input_image, target_class=None):
        """
        Generate Grad-CAM heatmap using simplified approach
        
        Args:
            input_image: numpy array (H, W, 3) BGR
            target_class: Not used in this simplified version
        
        Returns:
            cam: Grad-CAM heatmap (H, W)
        """
        try:
            # Reset
            self.gradients = None
            self.activations = None
            
            # Get device
            device = next(self.model.parameters()).device
            
            # Prepare image - use model's own preprocessing
            image_rgb = cv2.cvtColor(input_image, cv2.COLOR_BGR2RGB)
            
            # Use the model in eval mode but enable gradients
            self.model.eval()
            
            # Enable gradients for this operation
            image_tensor = torch.from_numpy(image_rgb).permute(2, 0, 1).float() / 255.0
            image_tensor = image_tensor.unsqueeze(0).to(device)
            image_tensor.requires_grad = True
            
            # Forward pass through wrapped model (stays in eval mode)
            # This goes through AutoShape -> DetectMultiBackend -> DetectionModel
            with torch.set_grad_enabled(True):
                # Call the model
                results = self.model.model(image_tensor)
                
                # Results is a tuple, get the first output
                if isinstance(results, tuple):
                    output = results[0]
                else:
                    output = results
                
                # Check if we captured activations
                if self.activations is None:
                    print("Warning: No activations captured, returning empty heatmap")
                    return np.zeros((input_image.shape[0], input_image.shape[1]))
                
                # Use sum of all outputs as target
                # This represents overall detection response
                target = output.sum()
                
                # Backward pass
                target.backward()
                
                # Compute Grad-CAM
                if self.gradients is not None:
                    # Get gradients and activations
                    gradients = self.gradients  # (B, C, H, W)
                    activations = self.activations  # (B, C, H, W)
                    
                    # Squeeze batch dimension if needed
                    if gradients.dim() == 4:
                        gradients = gradients.squeeze(0)  # (C, H, W)
                        activations = activations.squeeze(0)  # (C, H, W)
                    
                    # Global average pooling of gradients
                    weights = gradients.mean(dim=(1, 2), keepdim=True)  # (C, 1, 1)
                    
                    # Weighted combination
                    cam = (weights * activations).sum(dim=0)  # (H, W)
                    
                    # ReLU and normalize
                    cam = torch.relu(cam)
                    cam = cam.cpu().numpy()
                    
                    # Normalize to [0, 1]
                    if cam.max() > 0:
                        cam = cam / cam.max()
                    
                    # Resize to input image size
                    cam = cv2.resize(cam, (input_image.shape[1], input_image.shape[0]))
                    
                    return cam
                else:
                    print("Warning: No gradients captured")
                    return np.zeros((input_image.shape[0], input_image.shape[1]))
                    
        except Exception as e:
            print(f"Error generating Grad-CAM: {e}")
            import traceback
            traceback.print_exc()
            return np.zeros((input_image.shape[0], input_image.shape[1]))
    
    def remove_hooks(self):
        """Remove registered hooks"""
        for handle in self.hook_handles:
            handle.remove()


def get_target_layer(model):
    """
    Get the last convolutional layer from YOLOv5 model
    """
    # YOLOv5 v7.0 structure:
    # model (AutoShape) -> model.model (DetectMultiBackend) -> model.model.model (DetectionModel)
    
    try:
        # Navigate to DetectionModel
        detection_model = model.model.model
        
        # DetectionModel has a 'model' attribute which is a Sequential
        if hasattr(detection_model, 'model'):
            sequential_model = detection_model.model
            
            print(f"  Found {len(sequential_model)} layers")
            
            target_layer = sequential_model[10]
            
            return target_layer
        else:
            raise AttributeError("DetectionModel doesn't have expected structure")
            
    except Exception as e:
        print(f"  Error: {e}")
        print("  Trying alternative approach...")
        
        try:
            return model.model.model.model[10]
        except:
            print("  Could not find suitable layer")
            return None


print("✓ YOLOv5 Grad-CAM class defined")
```

```python
def generate_yolo_attention_map(image, model, detections, spread_factor=2.0, blur_size=31):
    """
    Generate attention map based on YOLOv5 detections with 2x spread
    
    Args:
        image: input image (H, W, 3)
        model: YOLOv5 model
        detections: list of detection dicts
        spread_factor: Spread multiplier (default 2.0 for smooth heatmap)
        blur_size: Gaussian blur kernel size (default 31)
    
    Returns:
        attention_map: heatmap showing detection areas (H, W)
    """
    h, w = image.shape[:2]
    attention_map = np.zeros((h, w), dtype=np.float32)
    
    # Create Gaussian heatmap for each detection
    for det in detections:
        x1, y1, x2, y2 = det['bbox'].astype(int)
        confidence = det['confidence']
        
        # Get center and size
        cx = int((x1 + x2) / 2)
        cy = int((y1 + y2) / 2)
        w_box = x2 - x1
        h_box = y2 - y1
        
        # Create 2D Gaussian with 2x spread
        sigma_x = (w_box / 4) * spread_factor
        sigma_y = (h_box / 4) * spread_factor
        
        # Expand the region
        pad_x = int(w_box * spread_factor)
        pad_y = int(h_box * spread_factor)
        
        # Create meshgrid for the detection region
        y_range = np.arange(max(0, y1-pad_y), min(h, y2+pad_y))
        x_range = np.arange(max(0, x1-pad_x), min(w, x2+pad_x))
        
        if len(y_range) > 0 and len(x_range) > 0:
            xx, yy = np.meshgrid(x_range, y_range)
            
            # Gaussian formula
            gaussian = np.exp(-(((xx - cx) ** 2) / (2 * sigma_x ** 2) + 
                              ((yy - cy) ** 2) / (2 * sigma_y ** 2)))
            
            # Weight by confidence
            gaussian = gaussian * confidence
            
            # Add to attention map
            y_start = max(0, y1-pad_y)
            y_end = min(h, y2+pad_y)
            x_start = max(0, x1-pad_x)
            x_end = min(w, x2+pad_x)
            
            attention_map[y_start:y_end, x_start:x_end] = np.maximum(
                attention_map[y_start:y_end, x_start:x_end],
                gaussian
            )
    
    # Apply Gaussian blur for smoothness
    attention_map = cv2.GaussianBlur(attention_map, (blur_size, blur_size), 0)
    
    # Normalize
    if attention_map.max() > 0:
        attention_map = attention_map / attention_map.max()
    
    return attention_map


def generate_saliency_map(image, model):
    """
    Generate saliency map using input gradients
    Shows which pixels are important for the model's predictions
    
    Args:
        image: numpy array (H, W, 3) BGR
        model: YOLOv5 model
    
    Returns:
        saliency: saliency map (H, W)
    """
    try:
        device = next(model.parameters()).device
        
        # Prepare image
        image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        image_tensor = torch.from_numpy(image_rgb).permute(2, 0, 1).float() / 255.0
        image_tensor = image_tensor.unsqueeze(0).to(device)
        image_tensor.requires_grad = True
        
        model.eval()
        
        # Forward pass
        with torch.set_grad_enabled(True):
            output = model(image_tensor)
            
            # Get predictions
            pred = output.xyxy[0]
            
            if len(pred) > 0:
                # Use sum of all confidences as target
                target = pred[:, 4].sum()
                
                # Backward
                target.backward()
                
                # Get gradients
                gradients = image_tensor.grad.data.abs()
                
                # Take max across color channels
                saliency = gradients.squeeze(0).max(dim=0)[0]
                saliency = saliency.cpu().numpy()
                
                # Normalize
                if saliency.max() > 0:
                    saliency = saliency / saliency.max()
                
                return saliency
            else:
                return np.zeros((image.shape[0], image.shape[1]))
                
    except Exception as e:
        print(f"Saliency map error: {e}")
        return np.zeros((image.shape[0], image.shape[1]))


# Test both methods on latest frame with 2x spread
if latest_image is not None:
    print("Generating visual explanations (2x Spread, 31px Blur)...")
    print("-" * 60)
    
    # Save current image
    current_image = latest_image.copy()
    
    # Get detections
    results, inf_time = detect_objects(current_image, models['yolov5s'])
    detections = parse_detections(results)
    
    print(f"Found {len(detections)} detections")
    
    if len(detections) > 0:
        # Method 1: Detection-based attention map (2x spread)
        print("\n1. Generating detection attention map (2x spread)...")
        attention_map = generate_yolo_attention_map(current_image, models['yolov5s'], detections)
        print("✓ Attention map generated")
        
        # Method 2: Saliency map
        print("2. Generating saliency map...")
        saliency_map = generate_saliency_map(current_image, models['yolov5s'])
        print("✓ Saliency map generated")
        
        # Visualize all
        fig, axes = plt.subplots(2, 2, figsize=(16, 14))
        
        # Original image
        axes[0, 0].imshow(cv2.cvtColor(current_image, cv2.COLOR_BGR2RGB))
        axes[0, 0].set_title("Original Image", fontsize=14)
        axes[0, 0].axis('off')
        
        # Attention heatmap
        axes[0, 1].imshow(attention_map, cmap='jet')
        axes[0, 1].set_title("Detection Attention Map (2x Spread)", fontsize=14)
        axes[0, 1].axis('off')
        
        # Saliency map
        axes[1, 0].imshow(saliency_map, cmap='jet')
        axes[1, 0].set_title("Saliency Map", fontsize=14)
        axes[1, 0].axis('off')
        
        # Overlay attention on image
        image_rgb = cv2.cvtColor(current_image, cv2.COLOR_BGR2RGB).astype(np.float32) / 255.0
        attention_colored = cv2.applyColorMap(np.uint8(255 * attention_map), cv2.COLORMAP_JET)
        attention_colored = cv2.cvtColor(attention_colored, cv2.COLOR_BGR2RGB).astype(np.float32) / 255.0
        overlay = 0.5 * image_rgb + 0.5 * attention_colored
        overlay = np.clip(overlay, 0, 1)
        
        axes[1, 1].imshow(overlay)
        axes[1, 1].set_title("Attention Overlay", fontsize=14)
        axes[1, 1].axis('off')
        
        plt.tight_layout()
        plt.show()
        
        print("-" * 60)
        print("✓ All visualizations complete!")
        print(f"\nExplanation methods:")
        print("  - Detection Attention (2x Spread): Shows smooth attention around detected objects")
        print("  - Saliency Map: Shows important pixels for predictions")
        
    else:
        print("⚠️ No detections found!")
        print("   Try adding traffic or drive to populated area")
        
        # Show image anyway
        plt.figure(figsize=(14, 10))
        plt.imshow(cv2.cvtColor(current_image, cv2.COLOR_BGR2RGB))
        plt.title("Current Frame (No Detections)")
        plt.axis('off')
        plt.show()
    
else:
    print("⚠️ No image available")
```

```python
# ========================================
# CELL 1: LiDAR Setup with Rolling Buffer
# ========================================

import numpy as np
import time
import threading
from collections import deque
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN

print("Setting up LiDAR sensor with rolling buffer...")

# Rolling buffer - always keeps last 10 frames
lidar_buffer = deque(maxlen=10)
buffer_lock = threading.Lock()

def lidar_callback(lidar_measurement):
    """Process and store LiDAR data in rolling buffer"""
    # Convert to numpy array
    points = np.frombuffer(lidar_measurement.raw_data, dtype=np.float32)
    points = np.reshape(points, (int(points.shape[0] / 4), 4))
    point_cloud = points[:, :3]  # x, y, z
    
    # Add to buffer (thread-safe)
    with buffer_lock:
        lidar_buffer.append(point_cloud.copy())

# Get LiDAR blueprint
blueprint_library = world.get_blueprint_library()
lidar_bp = blueprint_library.find('sensor.lidar.ray_cast')

# Configure LiDAR
lidar_bp.set_attribute('channels', '64')
lidar_bp.set_attribute('range', '80.0')
lidar_bp.set_attribute('points_per_second', '500000')
lidar_bp.set_attribute('rotation_frequency', '10')
lidar_bp.set_attribute('upper_fov', '10')
lidar_bp.set_attribute('lower_fov', '-25')

# Position on top of vehicle
lidar_transform = carla.Transform(
    carla.Location(x=0.0, y=0.0, z=2.5),
    carla.Rotation(pitch=0.0, yaw=0.0, roll=0.0)
)

# Spawn LiDAR
lidar_sensor = world.spawn_actor(lidar_bp, lidar_transform, attach_to=vehicle)
lidar_sensor.listen(lidar_callback)

print("✓ LiDAR sensor attached with rolling buffer")
print("  Buffer size: 10 frames (auto-updates)")
print("  Channels: 64")
print("  Range: 80m")
print("  Points/sec: 500,000")

# Wait for buffer to fill
print("\nFilling buffer...")
time.sleep(1.5)

with buffer_lock:
    print(f"✓ Buffer contains {len(lidar_buffer)} frames")
    if len(lidar_buffer) > 0:
        print(f"✓ Each frame has ~{len(lidar_buffer[0])} points")
```

```python
# ========================================
# CELL 2: Get 360° LiDAR Snapshot
# ========================================

def get_lidar_360():
    """
    Get complete 360° LiDAR data from buffer
    Returns accumulated point cloud from last 10 frames
    """
    with buffer_lock:
        if len(lidar_buffer) < 10:
            print(f"⚠️ Buffer only has {len(lidar_buffer)} frames, waiting...")
            return None
        
        # Combine all frames in buffer
        all_points = list(lidar_buffer)
    
    # Stack into single point cloud
    accumulated = np.vstack(all_points)
    
    return accumulated

print("✓ Snapshot function created")
print("\nUsage: snapshot = get_lidar_360()")
```

```python
# ========================================
# CELL 3 : Filter out street lights
# ========================================

def detect_vehicles_from_lidar(point_cloud, debug=False):
    """
    Detect vehicles with street light filtering
    Street lights: tall, thin, small footprint
    """
    if point_cloud is None or len(point_cloud) == 0:
        return []
    
    GROUND_Z = -2.5
    
    if debug:
        print(f"Total points: {len(point_cloud)}")
    
    # Filter above ground
    height_mask = (point_cloud[:, 2] > GROUND_Z + 0.3) & (point_cloud[:, 2] < GROUND_Z + 3.5)
    above_ground = point_cloud[height_mask]
    
    if debug:
        print(f"Above ground: {len(above_ground)} ({len(above_ground)/len(point_cloud)*100:.1f}%)")
    
    if len(above_ground) < 10:
        return []
    
    filtered = above_ground
    
    # Split into close and far ranges
    distances = np.sqrt(filtered[:, 0]**2 + filtered[:, 1]**2)
    
    close_range = filtered[distances < 10]
    far_range = filtered[distances >= 10]
    
    if debug:
        print(f"Close range (<10m): {len(close_range)} points")
        print(f"Far range (≥10m): {len(far_range)} points")
    
    vehicles = []
    
    # Process close range
    if len(close_range) >= 5:
        from sklearn.cluster import DBSCAN
        clustering_close = DBSCAN(eps=0.5, min_samples=5).fit(close_range[:, :2])
        labels_close = clustering_close.labels_
        
        unique_labels_close = set(labels_close)
        unique_labels_close.discard(-1)
        
        if debug:
            print(f"Close clusters (eps=0.5): {len(unique_labels_close)}")
        
        for label in unique_labels_close:
            cluster_points = close_range[labels_close == label]
            vehicles.extend(process_cluster(cluster_points, debug))
    
    # Process far range
    if len(far_range) >= 5:
        from sklearn.cluster import DBSCAN
        clustering_far = DBSCAN(eps=1.5, min_samples=5).fit(far_range[:, :2])
        labels_far = clustering_far.labels_
        
        unique_labels_far = set(labels_far)
        unique_labels_far.discard(-1)
        
        if debug:
            print(f"Far clusters (eps=1.5): {len(unique_labels_far)}")
        
        for label in unique_labels_far:
            cluster_points = far_range[labels_far == label]
            vehicles.extend(process_cluster(cluster_points, debug))
    
    vehicles.sort(key=lambda v: v['distance'])
    return vehicles


def process_cluster(cluster_points, debug=False):
    """
    Process cluster and filter out street lights/poles
    """
    if len(cluster_points) < 5:
        return []
    
    center_x = np.mean(cluster_points[:, 0])
    center_y = np.mean(cluster_points[:, 1])
    center_z = np.mean(cluster_points[:, 2])
    
    size_x = np.max(cluster_points[:, 0]) - np.min(cluster_points[:, 0])
    size_y = np.max(cluster_points[:, 1]) - np.min(cluster_points[:, 1])
    size_z = np.max(cluster_points[:, 2]) - np.min(cluster_points[:, 2])
    
    # Calculate footprint area (X × Y)
    footprint_area = size_x * size_y
    
    # STREET LIGHT FILTER
    # Street lights are: very tall (Z>2.5m), thin (small X/Y), small footprint
    is_street_light = (
        size_z > 2.5 and  # Tall
        footprint_area < 0.5  # Small footprint (less than 0.5 m²)
    )
    
    if is_street_light:
        if debug:
            print(f"  ✗ FILTERED: Street light/pole - footprint={footprint_area:.2f}m², height={size_z:.2f}m")
        return []
    
    # Size filters for vehicles
    if size_x < 0.05 or size_x > 10.0:
        return []
    if size_y < 0.3 or size_y > 4.0:
        return []
    if size_z < 0.3 or size_z > 4.0:
        return []
    
    # Aspect ratio check - vehicles should be wider than tall
    # Street lights: height >> width
    aspect_ratio = size_z / max(size_x, size_y)
    if aspect_ratio > 4.0:  # Much taller than wide
        if debug:
            print(f"  ✗ FILTERED: Vertical object - aspect_ratio={aspect_ratio:.1f}")
        return []
    
    distance = np.sqrt(center_x**2 + center_y**2)
    
    if distance > 30:
        return []
    
    vehicle = {
        'x': center_x,
        'y': center_y,
        'z': center_z,
        'distance': distance,
        'size_x': size_x,
        'size_y': size_y,
        'size_z': size_z,
        'num_points': len(cluster_points)
    }
    
    if debug:
        direction = ("FRONT" if center_x > 0 else "BACK") + ("-RIGHT" if center_y > 0 else "-LEFT")
        print(f"  ✓ {direction}: dist={distance:.1f}m, size={size_x:.2f}×{size_y:.2f}×{size_z:.2f}m, pts={len(cluster_points)}")
    
    return [vehicle]


print("✓ Detection updated with street light filtering")
print("  Filters:")
print("    - Footprint area < 0.5m² (thin poles)")
print("    - Height > 2.5m with small footprint")
print("    - Aspect ratio > 4.0 (much taller than wide)")
```

```python
# ========================================
# CELL: Feature Extraction for SHAP
# ========================================

def extract_lidar_features(detected_vehicles):
    """
    Extract features from LiDAR detections
    """
    if len(detected_vehicles) == 0:
        return {
            'num_vehicles': 0,
            'closest_front': 100.0,
            'closest_back': 100.0,
            'closest_left': 100.0,
            'closest_right': 100.0,
            'vehicles_front': 0,
            'vehicles_back': 0,
            'vehicles_left': 0,
            'vehicles_right': 0,
            'closest_vehicle': 100.0
        }
    
    # Initialize distances
    closest_front = 100.0
    closest_back = 100.0
    closest_left = 100.0
    closest_right = 100.0
    
    vehicles_front = 0
    vehicles_back = 0
    vehicles_left = 0
    vehicles_right = 0
    
    for v in detected_vehicles:
        dist = v['distance']
        
        # Determine zone
        if v['x'] > 0:  # Front
            vehicles_front += 1
            closest_front = min(closest_front, dist)
        else:  # Back
            vehicles_back += 1
            closest_back = min(closest_back, dist)
        
        if v['y'] > 0:  # Right
            vehicles_right += 1
            closest_right = min(closest_right, dist)
        else:  # Left
            vehicles_left += 1
            closest_left = min(closest_left, dist)
    
    return {
        'num_vehicles': len(detected_vehicles),
        'closest_front': closest_front,
        'closest_back': closest_back,
        'closest_left': closest_left,
        'closest_right': closest_right,
        'vehicles_front': vehicles_front,
        'vehicles_back': vehicles_back,
        'vehicles_left': vehicles_left,
        'vehicles_right': vehicles_right,
        'closest_vehicle': min([v['distance'] for v in detected_vehicles])
    }


def extract_vehicle_state(vehicle):
    """
    Extract current vehicle state
    """
    vel = vehicle.get_velocity()
    speed_kmh = 3.6 * np.sqrt(vel.x**2 + vel.y**2 + vel.z**2)
    
    control = vehicle.get_control()
    
    return {
        'speed': speed_kmh,
        'throttle': control.throttle,
        'brake': control.brake,
        'steering': control.steer
    }


def extract_features():
    """
    Extract all features for current frame
    """
    # Get LiDAR snapshot
    snapshot = get_lidar_360()
    detected = detect_vehicles_from_lidar(snapshot, debug=False)
    
    # Extract features
    lidar_features = extract_lidar_features(detected)
    vehicle_features = extract_vehicle_state(vehicle)
    
    # Combine
    features = {**lidar_features, **vehicle_features}
    
    # Add computed features
    safe_distance = 2.0 * features['speed'] / 3.6  # 2 seconds at current speed
    features['safe_distance_front'] = 1 if features['closest_front'] > safe_distance else 0
    
    return features


print("✓ Feature extraction functions created")

# Test it
test_features = extract_features()
print("\nCurrent features:")
for key, val in test_features.items():
    print(f"  {key}: {val:.2f}" if isinstance(val, float) else f"  {key}: {val}")
```

```python
# ========================================
# CELL: Data Collection for SHAP Training
# ========================================

import pandas as pd
from collections import deque

# Storage for training data
training_data = []
feature_buffer = deque(maxlen=5)  # Store last 5 frames for temporal context

def label_driving_action(features, prev_features=None):
    """
    Label the driving action based on features
    This simulates what the autopilot is doing
    """
    # Brake condition
    if features['closest_front'] < 10 and features['speed'] > 20:
        return 'brake'
    
    # Accelerate condition
    elif features['closest_front'] > 20 and features['speed'] < 40:
        return 'accelerate'
    
    # Maintain
    else:
        return 'maintain'


def collect_data_point():
    """
    Collect one data point (features + label)
    """
    features = extract_features()
    
    # Get previous features if available
    prev_features = feature_buffer[-1] if len(feature_buffer) > 0 else None
    
    # Label the action
    action = label_driving_action(features, prev_features)
    
    # Store features and action
    data_point = features.copy()
    data_point['action'] = action
    
    training_data.append(data_point)
    feature_buffer.append(features)
    
    return data_point


def collect_training_data(duration=30, sample_rate=0.1):
    """
    Collect training data for specified duration
    
    Args:
        duration: seconds to collect
        sample_rate: time between samples (seconds)
    """
    import time
    
    print(f"Collecting data for {duration} seconds...")
    print("Drive around naturally with autopilot!")
    
    start_time = time.time()
    samples = 0
    
    while time.time() - start_time < duration:
        data_point = collect_data_point()
        samples += 1
        
        if samples % 10 == 0:
            print(f"  Collected {samples} samples... Action: {data_point['action']}")
        
        time.sleep(sample_rate)
    
    print(f"\n✓ Collected {len(training_data)} total samples")
    
    # Show distribution
    df = pd.DataFrame(training_data)
    print("\nAction distribution:")
    print(df['action'].value_counts())
    
    return df


print("✓ Data collection functions created")
print("\nUsage:")
print("  df = collect_training_data(duration=30)  # Collect for 30 seconds")
print("  # OR manually collect points:")
print("  collect_data_point()  # Collect single point")
```

```python
# ========================================
# CELL: Train Model and Setup SHAP
# ========================================

import shap
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder

def train_explainer_model(df):
    """
    Train a model to predict driving actions and create SHAP explainer
    """
    print("Training explainer model...")
    
    # Separate features and labels
    feature_columns = [col for col in df.columns if col != 'action']
    X = df[feature_columns]
    y = df['action']
    
    # Encode labels
    label_encoder = LabelEncoder()
    y_encoded = label_encoder.fit_transform(y)
    
    print(f"\nFeatures: {len(feature_columns)}")
    print(f"Samples: {len(X)}")
    print(f"Classes: {label_encoder.classes_}")
    
    # Split data
    X_train, X_test, y_train, y_test = train_test_split(
        X, y_encoded, test_size=0.2, random_state=42
    )
    
    # Train Random Forest
    model = RandomForestClassifier(
        n_estimators=100,
        max_depth=10,
        random_state=42,
        n_jobs=-1
    )
    
    model.fit(X_train, y_train)
    
    # Evaluate
    train_score = model.score(X_train, y_train)
    test_score = model.score(X_test, y_test)
    
    print(f"\n✓ Model trained!")
    print(f"  Training accuracy: {train_score:.3f}")
    print(f"  Test accuracy: {test_score:.3f}")
    
    # Create SHAP explainer
    print("\nCreating SHAP explainer...")
    explainer = shap.TreeExplainer(model)
    
    print("✓ SHAP explainer ready!")
    
    return model, explainer, label_encoder, feature_columns


def explain_current_decision(model, explainer, label_encoder, feature_columns):
    """
    Explain the current driving decision using SHAP
    """
    # Get current features
    features = extract_features()
    
    # Convert to dataframe
    X = pd.DataFrame([features])[feature_columns]
    
    # Predict action
    prediction = model.predict(X)[0]
    action = label_encoder.inverse_transform([prediction])[0]
    
    # Get prediction probabilities
    probabilities = model.predict_proba(X)[0]
    
    # Calculate SHAP values
    shap_values = explainer.shap_values(X)
    
    # Display explanation
    print("="*60)
    print("  DRIVING DECISION EXPLANATION")
    print("="*60)
    print(f"\nPredicted Action: {action.upper()}")
    print(f"\nProbabilities:")
    for i, class_name in enumerate(label_encoder.classes_):
        print(f"  {class_name}: {probabilities[i]:.3f}")
    
    print(f"\nTop Contributing Features:")
    
    # Get SHAP values for predicted class
    shap_for_class = shap_values[prediction][0]
    
    # Get feature importance
    feature_importance = list(zip(feature_columns, shap_for_class))
    feature_importance.sort(key=lambda x: abs(x[1]), reverse=True)
    
    print(f"\n(For action: {action})")
    for i, (feat, val) in enumerate(feature_importance[:5], 1):
        direction = "→ INCREASES" if val > 0 else "→ DECREASES"
        print(f"  {i}. {feat:20s} = {features[feat]:6.2f}  |  Impact: {val:+.3f} {direction} likelihood")
    
    return action, shap_values, features


print("✓ SHAP functions created")
print("\nNext steps:")
print("1. Train model: model, explainer, encoder, features = train_explainer_model(df)")
print("2. Explain decision: explain_current_decision(model, explainer, encoder, features)")
```

```python
# ========================================
# CELL: SHAP Visualization
# ========================================

def visualize_shap_explanation(model, explainer, label_encoder, feature_columns, save_path=None):
    """
    Create comprehensive SHAP visualization for current decision
    """
    # Get current features
    features = extract_features()
    X = pd.DataFrame([features])[feature_columns]
    
    # Predict
    prediction = model.predict(X)[0]
    action = label_encoder.inverse_transform([prediction])[0]
    
    # Calculate SHAP values
    shap_values = explainer.shap_values(X)
    
    # Create figure with subplots
    fig = plt.figure(figsize=(20, 12), facecolor='white')
    
    # ============================================
    # 1. WATERFALL PLOT - Shows feature contributions
    # ============================================
    ax1 = plt.subplot(2, 2, 1)
    
    shap_for_class = shap_values[prediction][0]
    feature_vals = X.values[0]
    
    # Sort by absolute SHAP value
    indices = np.argsort(np.abs(shap_for_class))[::-1][:8]  # Top 8 features
    
    sorted_features = [feature_columns[i] for i in indices]
    sorted_shap = [shap_for_class[i] for i in indices]
    sorted_vals = [feature_vals[i] for i in indices]
    
    # Create waterfall
    colors = ['#ff4444' if s < 0 else '#44ff44' for s in sorted_shap]
    y_pos = np.arange(len(sorted_features))
    
    ax1.barh(y_pos, sorted_shap, color=colors, alpha=0.7, edgecolor='black', linewidth=2)
    ax1.set_yticks(y_pos)
    ax1.set_yticklabels([f"{feat}\n={val:.1f}" for feat, val in zip(sorted_features, sorted_vals)], fontsize=10)
    ax1.set_xlabel('SHAP Value (Impact on Decision)', fontsize=12, weight='bold')
    ax1.set_title(f'Feature Contributions for: {action.upper()}', fontsize=14, weight='bold')
    ax1.axvline(x=0, color='black', linestyle='--', linewidth=2)
    ax1.grid(True, alpha=0.3)
    
    # ============================================
    # 2. SUMMARY - Feature values and importance
    # ============================================
    ax2 = plt.subplot(2, 2, 2)
    
    summary_text = f"DECISION EXPLANATION\n{'='*40}\n\n"
    summary_text += f"Action: {action.upper()}\n\n"
    summary_text += f"Key Factors:\n\n"
    
    for i, (feat, shap_val, val) in enumerate(zip(sorted_features[:5], sorted_shap[:5], sorted_vals[:5]), 1):
        direction = "INCREASES" if shap_val > 0 else "DECREASES"
        summary_text += f"{i}. {feat}\n"
        summary_text += f"   Value: {val:.2f}\n"
        summary_text += f"   Impact: {direction} {action}\n"
        summary_text += f"   Strength: {abs(shap_val):.3f}\n\n"
    
    ax2.text(0.05, 0.95, summary_text, transform=ax2.transAxes,
            fontsize=11, verticalalignment='top', family='monospace',
            bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))
    ax2.axis('off')
    
    # ============================================
    # 3. FORCE PLOT - Base value to prediction
    # ============================================
    ax3 = plt.subplot(2, 2, 3)
    
    base_value = explainer.expected_value[prediction]
    
    # Cumulative sum for force plot
    cumsum = np.cumsum([base_value] + list(sorted_shap))
    
    ax3.plot(range(len(cumsum)), cumsum, 'o-', linewidth=3, markersize=10, color='#4444ff')
    ax3.axhline(y=base_value, color='gray', linestyle='--', label='Base Value')
    ax3.set_xlabel('Feature Added', fontsize=12, weight='bold')
    ax3.set_ylabel('Model Output', fontsize=12, weight='bold')
    ax3.set_title('Sequential Feature Impact', fontsize=14, weight='bold')
    ax3.grid(True, alpha=0.3)
    ax3.legend()
    
    # ============================================
    # 4. FEATURE VALUES - Current situation
    # ============================================
    ax4 = plt.subplot(2, 2, 4)
    
    situation_text = f"CURRENT SITUATION\n{'='*40}\n\n"
    situation_text += f"Vehicle State:\n"
    situation_text += f"  Speed: {features['speed']:.1f} km/h\n"
    situation_text += f"  Throttle: {features['throttle']:.2f}\n"
    situation_text += f"  Brake: {features['brake']:.2f}\n\n"
    
    situation_text += f"Environment:\n"
    situation_text += f"  Vehicles detected: {features['num_vehicles']}\n"
    situation_text += f"  Closest (front): {features['closest_front']:.1f}m\n"
    situation_text += f"  Closest (back): {features['closest_back']:.1f}m\n"
    situation_text += f"  Closest (left): {features['closest_left']:.1f}m\n"
    situation_text += f"  Closest (right): {features['closest_right']:.1f}m\n\n"
    
    situation_text += f"Safety:\n"
    situation_text += f"  Safe distance: {'YES' if features['safe_distance_front'] else 'NO'}\n"
    
    ax4.text(0.05, 0.95, situation_text, transform=ax4.transAxes,
            fontsize=11, verticalalignment='top', family='monospace',
            bbox=dict(boxstyle='round', facecolor='lightblue', alpha=0.5))
    ax4.axis('off')
    
    plt.suptitle(f'SHAP Explanation: {action.upper()} Decision', 
                fontsize=18, weight='bold', y=0.98)
    
    plt.tight_layout()
    
    if save_path:
        plt.savefig(save_path, dpi=150, bbox_inches='tight')
        print(f"✓ Saved to {save_path}")
    
    plt.show()
    
    return action, shap_values


print("✓ SHAP visualization function created")
print("\nUsage:")
print("  visualize_shap_explanation(model, explainer, label_encoder, feature_columns)")
print("  # Or save to file:")
print("  visualize_shap_explanation(model, explainer, label_encoder, feature_columns, 'explanation.png')")
```

```python
# ========================================
# CELL: Real-Time XAI Dashboard
# ========================================

from matplotlib.gridspec import GridSpec
import matplotlib.patches as mpatches
from datetime import datetime

class XAIDashboard:
    """
    Real-time explainable AI dashboard for autonomous driving
    """
    
    def __init__(self, vehicle, yolo_model, shap_model, shap_explainer, label_encoder, feature_columns):
        self.vehicle = vehicle
        self.yolo_model = yolo_model
        self.shap_model = shap_model
        self.shap_explainer = shap_explainer
        self.label_encoder = label_encoder
        self.feature_columns = feature_columns
        
        # History tracking
        self.action_history = deque(maxlen=20)
        self.frame_count = 0
        
        # Setup figure
        self.setup_figure()
    
    def setup_figure(self):
        """
        Create the dashboard layout
        """
        self.fig = plt.figure(figsize=(24, 14), facecolor='#0a0a0a')
        
        # Create grid: 2 rows, 3 columns
        gs = GridSpec(2, 3, figure=self.fig, hspace=0.3, wspace=0.3,
                     left=0.05, right=0.95, top=0.95, bottom=0.05)
        
        # Panel 1: LiDAR (top left)
        self.ax_lidar = self.fig.add_subplot(gs[0, 0])
        self.ax_lidar.set_facecolor('#0a0a0a')
        
        # Panel 2: Camera + YOLO (top middle)
        self.ax_camera = self.fig.add_subplot(gs[0, 1])
        self.ax_camera.set_facecolor('#0a0a0a')
        
        # Panel 3: Attention Map (top right)
        self.ax_attention = self.fig.add_subplot(gs[0, 2])
        self.ax_attention.set_facecolor('#0a0a0a')
        
        # Panel 4: SHAP (bottom left)
        self.ax_shap = self.fig.add_subplot(gs[1, 0])
        self.ax_shap.set_facecolor('#0a0a0a')
        
        # Panel 5: Decision (bottom middle)
        self.ax_decision = self.fig.add_subplot(gs[1, 1])
        self.ax_decision.set_facecolor('#0a0a0a')
        
        # Panel 6: Context (bottom right)
        self.ax_context = self.fig.add_subplot(gs[1, 2])
        self.ax_context.set_facecolor('#0a0a0a')
    
    def update_lidar_panel(self):
        """
        Update LiDAR bird's eye view
        """
        self.ax_lidar.clear()
        self.ax_lidar.set_facecolor('#0a0a0a')
        
        # Get LiDAR data
        snapshot = get_lidar_360()
        if snapshot is None:
            return
        
        GROUND_Z = -2.5
        above_ground = snapshot[(snapshot[:, 2] > GROUND_Z + 0.2) & (snapshot[:, 2] < GROUND_Z + 3.5)]
        
        # Plot point cloud
        if len(above_ground) > 0:
            heights = above_ground[:, 2] - GROUND_Z
            colors = plt.cm.plasma(np.clip(heights / 3.0, 0, 1))
            self.ax_lidar.scatter(above_ground[:, 1], above_ground[:, 0],
                                 c=colors, s=3, alpha=0.6, edgecolors='none')
        
        # Red arrow ego
        arrow = mpatches.Polygon([[0, 2.2], [-1.5, -2], [0, -1.5], [1.5, -2]],
                                fill=True, facecolor='#ff3333', edgecolor='#ff0000', linewidth=2)
        self.ax_lidar.add_patch(arrow)
        
        # Detect vehicles
        detected = detect_vehicles_from_lidar(snapshot, debug=False)
        
        # Draw vehicles
        for i, v in enumerate(detected, 1):
            display_x = max(v['size_x'], 3.0)
            display_y = max(v['size_y'], 1.5)
            
            distance_ratio = min(v['distance'] / 30.0, 1.0)
            box_color = plt.cm.RdYlGn_r(distance_ratio)
            
            rect = mpatches.Rectangle((v['y'] - display_y/2, v['x'] - display_x/2),
                                     display_y, display_x, fill=False, 
                                     edgecolor=box_color, linewidth=2, alpha=0.8)
            self.ax_lidar.add_patch(rect)
            
            # Label
            direction = "F" if v['x'] > 0 else "B"
            if abs(v['y']) > 0.5:
                direction += "R" if v['y'] > 0 else "L"
            
            self.ax_lidar.text(v['y'], v['x'] + display_x/2 + 0.5,
                              f"{direction}\n{v['distance']:.0f}m",
                              color='white', fontsize=7, ha='center', weight='bold')
        
        # Distance circles
        for r in [10, 20, 30]:
            circle = plt.Circle((0, 0), r, fill=False, edgecolor='#2a4a6a',
                               linestyle='--', linewidth=1, alpha=0.4)
            self.ax_lidar.add_patch(circle)
        
        self.ax_lidar.set_xlim(-35, 35)
        self.ax_lidar.set_ylim(-35, 35)
        self.ax_lidar.set_aspect('equal')
        self.ax_lidar.set_title('LiDAR 360°', color='white', fontsize=12, weight='bold')
        self.ax_lidar.tick_params(colors='white', labelsize=8)
        
        return detected
    
    def update_camera_panel(self):
        """
        Update camera view with YOLO detections
        """
        self.ax_camera.clear()
        self.ax_camera.set_facecolor('#0a0a0a')
        
        # Get camera image
        if latest_image is None:
            self.ax_camera.text(0.5, 0.5, 'No Camera', ha='center', va='center',
                               color='white', fontsize=14, transform=self.ax_camera.transAxes)
            return None
        
        image = latest_image.copy()
        
        # Run YOLO
        results = self.yolo_model(image)
        detections = parse_detections(results)
        
        # Draw bounding boxes
        for det in detections:
            x1, y1, x2, y2 = det['bbox'].astype(int)
            label = det['label']
            conf = det['confidence']
            
            color_map = {
                'person': (255, 0, 0),
                'car': (0, 255, 0),
                'truck': (0, 255, 255),
                'traffic light': (255, 255, 0)
            }
            color = color_map.get(label, (128, 128, 128))
            
            cv2.rectangle(image, (x1, y1), (x2, y2), color, 2)
            cv2.putText(image, f'{label} {conf:.2f}', (x1, y1-5),
                       cv2.FONT_HERSHEY_SIMPLEX, 0.5, color, 2)
        
        # Display
        self.ax_camera.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
        self.ax_camera.set_title(f'Camera + YOLO ({len(detections)} objects)', 
                                color='white', fontsize=12, weight='bold')
        self.ax_camera.axis('off')
        
        return detections
    
    def update_attention_panel(self, camera_detections):
        """
        Update attention map
        """
        self.ax_attention.clear()
        self.ax_attention.set_facecolor('#0a0a0a')
        
        if latest_image is None or camera_detections is None or len(camera_detections) == 0:
            self.ax_attention.text(0.5, 0.5, 'No Detections', ha='center', va='center',
                                  color='white', fontsize=14, transform=self.ax_attention.transAxes)
            return
        
        # Generate attention map
        attention_map = generate_yolo_attention_map(latest_image, self.yolo_model, camera_detections)
        
        # Apply blur for spread
        attention_map_blurred = cv2.GaussianBlur(attention_map, (31, 31), 0)
        
        # Overlay
        image_rgb = cv2.cvtColor(latest_image, cv2.COLOR_BGR2RGB)
        image_normalized = image_rgb.astype(np.float32) / 255.0
        
        heatmap_colored = plt.cm.hot(attention_map_blurred)[:, :, :3]
        overlay = 0.6 * image_normalized + 0.4 * heatmap_colored
        overlay = np.clip(overlay, 0, 1)
        
        # Display
        self.ax_attention.imshow(overlay)
        self.ax_attention.set_title('Attention Map', color='white', fontsize=12, weight='bold')
        self.ax_attention.axis('off')
    
    def update_shap_panel(self, features, shap_values, action):
        """
        Update SHAP feature importance
        """
        self.ax_shap.clear()
        self.ax_shap.set_facecolor('#0a0a0a')
        
        # Get prediction class index
        prediction = self.label_encoder.transform([action])[0]
        
        # Get SHAP values for this prediction
        shap_for_class = shap_values[prediction][0]
        
        # Top 8 features
        indices = np.argsort(np.abs(shap_for_class))[::-1][:8]
        
        top_features = [self.feature_columns[i] for i in indices]
        top_shap = [shap_for_class[i] for i in indices]
        top_vals = [features[feat] for feat in top_features]
        
        # Plot
        colors = ['#ff4444' if s < 0 else '#44ff44' for s in top_shap]
        y_pos = np.arange(len(top_features))
        
        self.ax_shap.barh(y_pos, top_shap, color=colors, alpha=0.8, edgecolor='white', linewidth=1)
        self.ax_shap.set_yticks(y_pos)
        self.ax_shap.set_yticklabels([f"{feat}\n={val:.1f}" for feat, val in zip(top_features, top_vals)], 
                                     fontsize=9, color='white')
        self.ax_shap.set_xlabel('Impact on Decision', fontsize=10, weight='bold', color='white')
        self.ax_shap.set_title(f'SHAP: {action.upper()}', color='white', fontsize=12, weight='bold')
        self.ax_shap.axvline(x=0, color='white', linestyle='--', linewidth=1)
        self.ax_shap.tick_params(colors='white', labelsize=8)
        self.ax_shap.grid(True, alpha=0.2, color='gray')
    
    def update_decision_panel(self, action, probabilities):
        """
        Update decision display
        """
        self.ax_decision.clear()
        self.ax_decision.set_facecolor('#0a0a0a')
        
        # Big action text
        action_colors = {
            'brake': '#ff4444',
            'accelerate': '#44ff44',
            'maintain': '#ffaa00'
        }
        color = action_colors.get(action, '#ffffff')
        
        self.ax_decision.text(0.5, 0.7, action.upper(), 
                             ha='center', va='center', fontsize=36, weight='bold',
                             color=color, transform=self.ax_decision.transAxes)
        
        # Confidence
        max_prob = max(probabilities)
        self.ax_decision.text(0.5, 0.5, f'Confidence: {max_prob:.1%}',
                             ha='center', va='center', fontsize=14, color='white',
                             transform=self.ax_decision.transAxes)
        
        # Action history timeline
        self.action_history.append(action)
        
        if len(self.action_history) > 1:
            history_y = 0.2
            history_x = np.linspace(0.1, 0.9, len(self.action_history))
            
            for x, hist_action in zip(history_x, self.action_history):
                hist_color = action_colors.get(hist_action, '#ffffff')
                self.ax_decision.plot(x, history_y, 'o', color=hist_color, markersize=8,
                                     transform=self.ax_decision.transAxes)
        
        self.ax_decision.axis('off')
    
    def update_context_panel(self, action, features, lidar_detections, camera_detections):
        """
        Update natural language explanation
        """
        self.ax_context.clear()
        self.ax_context.set_facecolor('#0a0a0a')
        
        # Generate explanation
        explanation = f"DECISION: {action.upper()}\n"
        explanation += "="*40 + "\n\n"
        explanation += "Why:\n"
        
        if action == 'brake':
            if features['closest_front'] < 15:
                explanation += f"• Vehicle {features['closest_front']:.1f}m ahead (CLOSE!)\n"
            if features['speed'] > 30:
                explanation += f"• High speed ({features['speed']:.1f} km/h)\n"
            if len(lidar_detections) > 3:
                explanation += f"• {len(lidar_detections)} vehicles detected\n"
        
        elif action == 'accelerate':
            if features['closest_front'] > 20:
                explanation += f"• Road clear ({features['closest_front']:.1f}m ahead)\n"
            if features['speed'] < 30:
                explanation += f"• Low speed ({features['speed']:.1f} km/h)\n"
            if features['safe_distance_front']:
                explanation += "• Safe distance maintained\n"
        
        else:  # maintain
            explanation += f"• Stable conditions\n"
            explanation += f"• Speed: {features['speed']:.1f} km/h\n"
            explanation += f"• Distance ahead: {features['closest_front']:.1f}m\n"
        
        explanation += f"\nDetections:\n"
        explanation += f"  LiDAR: {len(lidar_detections)} vehicles\n"
        explanation += f"  Camera: {len(camera_detections) if camera_detections else 0} objects\n"
        
        explanation += f"\n{datetime.now().strftime('%H:%M:%S')}"
        explanation += f" | Frame {self.frame_count}"
        
        # Display
        self.ax_context.text(0.05, 0.95, explanation, 
                            transform=self.ax_context.transAxes,
                            fontsize=10, verticalalignment='top', 
                            family='monospace', color='white',
                            bbox=dict(boxstyle='round', facecolor='#1a1a1a', 
                                    edgecolor='#00ff00', linewidth=2, alpha=0.9))
        self.ax_context.axis('off')
    
    def update(self):
        """
        Update entire dashboard
        """
        self.frame_count += 1
        
        try:
            # 1. Update LiDAR
            lidar_detections = self.update_lidar_panel()
            
            # 2. Update Camera + YOLO
            camera_detections = self.update_camera_panel()
            
            # 3. Update Attention Map
            self.update_attention_panel(camera_detections)
            
            # 4. Extract features and get SHAP
            features = extract_features()
            X = pd.DataFrame([features])[self.feature_columns]
            
            prediction = self.shap_model.predict(X)[0]
            probabilities = self.shap_model.predict_proba(X)[0]
            action = self.label_encoder.inverse_transform([prediction])[0]
            
            shap_values = self.shap_explainer.shap_values(X)
            
            # 5. Update SHAP
            self.update_shap_panel(features, shap_values, action)
            
            # 6. Update Decision
            self.update_decision_panel(action, probabilities)
            
            # 7. Update Context
            self.update_context_panel(action, features, lidar_detections, camera_detections)
            
            # Main title
            self.fig.suptitle('Explainable AI Dashboard - Autonomous Driving', 
                            color='white', fontsize=16, weight='bold')
            
            plt.draw()
            plt.pause(0.01)
            
            return True
            
        except Exception as e:
            print(f"Error updating dashboard: {e}")
            return False


print("✓ XAI Dashboard class created")
```

```python
# ========================================
# FIX: Parse YOLO Detections
# ========================================

def parse_detections(results):
    """
    Parse YOLOv5 results into list of detections
    """
    detections = []
    
    try:
        # Get predictions
        pred = results.xyxy[0].cpu().numpy()
        
        for detection in pred:
            x1, y1, x2, y2, conf, cls = detection
            
            # Get class name
            class_id = int(cls)
            class_name = results.names[class_id]
            
            detections.append({
                'bbox': np.array([x1, y1, x2, y2]),
                'confidence': float(conf),
                'label': class_name,
                'class_id': class_id
            })
    
    except Exception as e:
        print(f"Warning: Could not parse detections - {e}")
        return []
    
    return detections

print("✓ Fixed parse_detections function")
```

```python
# ========================================
# 4-Panel Dashboard: LiDAR | Attention | SHAP | Decision+Info
# ========================================

class XAIDashboard:
    """
    Real-time XAI dashboard - 4 PANELS
    """
    
    def __init__(self, vehicle, yolo_model, shap_model, shap_explainer, label_encoder, feature_columns):
        self.vehicle = vehicle
        self.yolo_model = yolo_model
        self.shap_model = shap_model
        self.shap_explainer = shap_explainer
        self.label_encoder = label_encoder
        self.feature_columns = feature_columns
        
        self.action_history = deque(maxlen=20)
        self.frame_count = 0
        
        self.setup_figure()
    
    def setup_figure(self):
        """
        Create 4-panel dashboard (2x2 grid)
        """
        self.fig = plt.figure(figsize=(15, 8), facecolor='#0a0a0a', dpi=100)
        
        # 2x2 grid
        gs = GridSpec(2, 2, figure=self.fig, 
                     hspace=0.3,
                     wspace=0.3,
                     left=0.05, right=0.97, 
                     top=0.94, bottom=0.06)
        
        # Panel 1: LiDAR (top left)
        self.ax_lidar = self.fig.add_subplot(gs[0, 0])
        self.ax_lidar.set_facecolor('#0a0a0a')
        
        # Panel 2: Attention Map (top right)
        self.ax_attention = self.fig.add_subplot(gs[0, 1])
        self.ax_attention.set_facecolor('#0a0a0a')
        
        # Panel 3: SHAP (bottom left)
        self.ax_shap = self.fig.add_subplot(gs[1, 0])
        self.ax_shap.set_facecolor('#0a0a0a')
        
        # Panel 4: Decision + Info COMBINED (bottom right)
        self.ax_decision = self.fig.add_subplot(gs[1, 1])
        self.ax_decision.set_facecolor('#0a0a0a')
    
    def update_lidar_panel(self):
        """LiDAR 360°"""
        self.ax_lidar.clear()
        self.ax_lidar.set_facecolor('#0a0a0a')
        
        snapshot = get_lidar_360()
        if snapshot is None:
            return []
        
        GROUND_Z = -2.5
        above_ground = snapshot[(snapshot[:, 2] > GROUND_Z + 0.2) & (snapshot[:, 2] < GROUND_Z + 3.5)]
        
        if len(above_ground) > 0:
            heights = above_ground[:, 2] - GROUND_Z
            colors = plt.cm.plasma(np.clip(heights / 3.0, 0, 1))
            self.ax_lidar.scatter(above_ground[:, 1], above_ground[:, 0],
                                 c=colors, s=2, alpha=0.5, edgecolors='none')
        
        # Red arrow ego
        arrow = mpatches.Polygon([[0, 2.2], [-1.5, -2], [0, -1.5], [1.5, -2]],
                                fill=True, facecolor='#ff3333', edgecolor='#ff0000', linewidth=2)
        self.ax_lidar.add_patch(arrow)
        
        detected = detect_vehicles_from_lidar(snapshot, debug=False)
        
        # Draw vehicles
        for i, v in enumerate(detected, 1):
            display_x = max(v['size_x'], 3.0)
            display_y = max(v['size_y'], 1.5)
            
            distance_ratio = min(v['distance'] / 30.0, 1.0)
            box_color = plt.cm.RdYlGn_r(distance_ratio)
            
            rect = mpatches.Rectangle((v['y'] - display_y/2, v['x'] - display_x/2),
                                     display_y, display_x, fill=False, 
                                     edgecolor=box_color, linewidth=2)
            self.ax_lidar.add_patch(rect)
            
            # Labels
            direction = "F" if v['x'] > 0 else "B"
            if abs(v['y']) > 0.5:
                direction += "R" if v['y'] > 0 else "L"
            
            self.ax_lidar.text(v['y'], v['x'] + display_x/2 + 0.5,
                              f"{direction}\n{v['distance']:.0f}m",
                              color='white', fontsize=8, ha='center', weight='bold')
        
        # Distance circles
        for r in [10, 20, 30]:
            circle = plt.Circle((0, 0), r, fill=False, edgecolor='#2a4a6a',
                               linestyle='--', linewidth=1, alpha=0.4)
            self.ax_lidar.add_patch(circle)
        
        self.ax_lidar.set_xlim(-35, 35)
        self.ax_lidar.set_ylim(-35, 35)
        self.ax_lidar.set_aspect('equal')
        self.ax_lidar.set_title('LiDAR 360°', color='white', fontsize=12, weight='bold')
        self.ax_lidar.tick_params(colors='white', labelsize=8)
        
        return detected
    
    def update_camera_panel(self):
        """Get camera detections (no display)"""
        if latest_image is None:
            return None
        
        results = self.yolo_model(latest_image)
        detections = parse_detections(results)
        
        return detections
    

    def update_attention_panel(self, camera_detections):
        """Attention Map with BLUE colormap"""
        self.ax_attention.clear()
        self.ax_attention.set_facecolor('#0a0a0a')
    
        if latest_image is None or camera_detections is None or len(camera_detections) == 0:
        self.ax_attention.text(0.5, 0.5, 'No Detections', ha='center', va='center',
                              color='white', fontsize=14, transform=self.ax_attention.transAxes)
        return
    
    attention_map = generate_yolo_attention_map(latest_image, self.yolo_model, camera_detections)
    attention_map_blurred = cv2.GaussianBlur(attention_map, (31, 31), 0)
    
    image_rgb = cv2.cvtColor(latest_image, cv2.COLOR_BGR2RGB)
    image_normalized = image_rgb.astype(np.float32) / 255.0
    
    # CHANGED: Use 'jet' colormap instead of 'hot' for blue tones
    heatmap_colored = plt.cm.jet(attention_map_blurred)[:, :, :3]
    # Or try 'cool' for pure blue: plt.cm.cool(attention_map_blurred)[:, :, :3]
    
    overlay = 0.6 * image_normalized + 0.4 * heatmap_colored
    overlay = np.clip(overlay, 0, 1)
    
    self.ax_attention.imshow(overlay)
    self.ax_attention.set_title('Attention Map', color='white', fontsize=12, weight='bold')
    self.ax_attention.axis('off')
    
    
    def update_shap_panel(self, features, shap_values, action):
        """SHAP Feature Importance"""
        self.ax_shap.clear()
        self.ax_shap.set_facecolor('#0a0a0a')
        
        prediction = self.label_encoder.transform([action])[0]
        shap_for_class = shap_values[prediction][0]
        
        # Top 6 features
        indices = np.argsort(np.abs(shap_for_class))[::-1][:6]
        
        top_features = [self.feature_columns[i] for i in indices]
        top_shap = [shap_for_class[i] for i in indices]
        top_vals = [features[feat] for feat in top_features]
        
        colors = ['#ff4444' if s < 0 else '#44ff44' for s in top_shap]
        y_pos = np.arange(len(top_features))
        
        self.ax_shap.barh(y_pos, top_shap, color=colors, alpha=0.8, edgecolor='white', linewidth=1)
        self.ax_shap.set_yticks(y_pos)
        
        # Feature labels
        labels = [f"{feat}\n={val:.1f}" for feat, val in zip(top_features, top_vals)]
        self.ax_shap.set_yticklabels(labels, fontsize=9, color='white')
        
        self.ax_shap.set_xlabel('Impact on Decision', fontsize=10, weight='bold', color='white')
        self.ax_shap.set_title(f'SHAP Explanation: {action.upper()}', color='white', fontsize=12, weight='bold')
        self.ax_shap.axvline(x=0, color='white', linestyle='--', linewidth=1)
        self.ax_shap.tick_params(colors='white', labelsize=8)
        self.ax_shap.grid(True, alpha=0.2, color='gray')
    
    def update_decision_panel(self, action, probabilities, features, lidar_detections, camera_detections):
        """Decision + Info COMBINED"""
        self.ax_decision.clear()
        self.ax_decision.set_facecolor('#0a0a0a')
        
        action_colors = {
            'brake': '#ff4444',
            'accelerate': '#44ff44',
            'maintain': '#ffaa00'
        }
        color = action_colors.get(action, '#ffffff')
        
        # BIG ACTION at top
        self.ax_decision.text(0.5, 0.85, action.upper(), 
                             ha='center', va='center', fontsize=32, weight='bold',
                             color=color, transform=self.ax_decision.transAxes)
        
        # Confidence
        max_prob = max(probabilities)
        self.ax_decision.text(0.5, 0.70, f'Confidence: {max_prob:.1%}',
                             ha='center', va='center', fontsize=14, color='white',
                             transform=self.ax_decision.transAxes)
        
        # Separator line
        self.ax_decision.plot([0.1, 0.9], [0.62, 0.62], color='white', linewidth=1, 
                             transform=self.ax_decision.transAxes)
        
        # INFO SECTION
        info_text = "SITUATION\n"
        info_text += "=" * 30 + "\n\n"
        info_text += f"Speed: {features['speed']:.1f} km/h\n"
        info_text += f"Closest Front: {features['closest_front']:.1f}m\n"
        info_text += f"Closest Back: {features['closest_back']:.1f}m\n"
        info_text += f"Safe Distance: {'YES' if features['safe_distance_front'] else 'NO'}\n\n"
        info_text += f"Detections:\n"
        info_text += f"  LiDAR: {len(lidar_detections)} vehicles\n"
        info_text += f"  Camera: {len(camera_detections) if camera_detections else 0} objects\n\n"
        
        # Why decision
        info_text += "Reasoning:\n"
        if action == 'brake':
            if features['closest_front'] < 15:
                info_text += f"• Close vehicle ({features['closest_front']:.1f}m)\n"
            if features['speed'] > 30:
                info_text += f"• High speed ({features['speed']:.1f} km/h)\n"
        elif action == 'accelerate':
            if features['closest_front'] > 20:
                info_text += f"• Clear ahead ({features['closest_front']:.1f}m)\n"
            if features['speed'] < 30:
                info_text += f"• Low speed ({features['speed']:.1f} km/h)\n"
        else:
            info_text += "• Maintaining current state\n"
        
        info_text += f"\n{datetime.now().strftime('%H:%M:%S')} | Frame {self.frame_count}"
        
        # Display info
        self.ax_decision.text(0.05, 0.55, info_text, 
                            transform=self.ax_decision.transAxes,
                            fontsize=9, verticalalignment='top', 
                            family='monospace', color='white',
                            bbox=dict(boxstyle='round', facecolor='#1a1a1a', 
                                    edgecolor='#00ff00', linewidth=2, alpha=0.9))
        
        # Action history timeline at bottom
        self.action_history.append(action)
        
        if len(self.action_history) > 1:
            history_y = 0.05
            history_x = np.linspace(0.1, 0.9, len(self.action_history))
            
            for x, hist_action in zip(history_x, self.action_history):
                hist_color = action_colors.get(hist_action, '#ffffff')
                self.ax_decision.plot(x, history_y, 'o', color=hist_color, markersize=8,
                                     transform=self.ax_decision.transAxes)
        
        self.ax_decision.axis('off')
    
    def update(self):
        """Update entire dashboard"""
        self.frame_count += 1
        
        try:
            # Get data
            lidar_detections = self.update_lidar_panel()
            camera_detections = self.update_camera_panel()
            self.update_attention_panel(camera_detections)
            
            # Extract features
            features = extract_features()
            X = pd.DataFrame([features])[self.feature_columns]
            
            # Predict action
            prediction = self.shap_model.predict(X)[0]
            probabilities = self.shap_model.predict_proba(X)[0]
            action = self.label_encoder.inverse_transform([prediction])[0]
            
            # SHAP
            shap_values = self.shap_explainer.shap_values(X)
            
            # Update panels
            self.update_shap_panel(features, shap_values, action)
            self.update_decision_panel(action, probabilities, features, lidar_detections, camera_detections)
            
            # Title
            self.fig.suptitle('Explainable AI Dashboard - Autonomous Driving', 
                            color='white', fontsize=14, weight='bold')
            
            plt.draw()
            plt.pause(0.001)
            
            return True
            
        except Exception as e:
            print(f"Error: {e}")
            return False


print("✓ 4-Panel Dashboard Created!")
print("  Layout: 2x2 grid")
print("  Panels: LiDAR | Attention | SHAP | Decision+Info")
```

```python
# ========================================
# 4-Panel Dashboard - ZOOMED LiDAR (20m)
# ========================================

class XAIDashboard:
    """
    Real-time XAI dashboard - 4 PANELS with ZOOMED LiDAR
    """
    
    def __init__(self, vehicle, yolo_model, shap_model, shap_explainer, label_encoder, feature_columns):
        self.vehicle = vehicle
        self.yolo_model = yolo_model
        self.shap_model = shap_model
        self.shap_explainer = shap_explainer
        self.label_encoder = label_encoder
        self.feature_columns = feature_columns
        
        self.action_history = deque(maxlen=20)
        self.frame_count = 0
        
        self.setup_figure()
    
    def setup_figure(self):
        """
        Create 4-panel dashboard (2x2 grid)
        """
        self.fig = plt.figure(figsize=(15, 8), facecolor='#0a0a0a', dpi=100)
        
        # 2x2 grid
        gs = GridSpec(2, 2, figure=self.fig, 
                     hspace=0.3,
                     wspace=0.3,
                     left=0.05, right=0.97, 
                     top=0.94, bottom=0.06)
        
        # Panel 1: LiDAR (top left)
        self.ax_lidar = self.fig.add_subplot(gs[0, 0])
        self.ax_lidar.set_facecolor('#0a0a0a')
        
        # Panel 2: Grad-CAM (top right)
        self.ax_attention = self.fig.add_subplot(gs[0, 1])
        self.ax_attention.set_facecolor('#0a0a0a')
        
        # Panel 3: SHAP (bottom left)
        self.ax_shap = self.fig.add_subplot(gs[1, 0])
        self.ax_shap.set_facecolor('#0a0a0a')
        
        # Panel 4: Decision + Info COMBINED (bottom right)
        self.ax_decision = self.fig.add_subplot(gs[1, 1])
        self.ax_decision.set_facecolor('#0a0a0a')
    
    def update_lidar_panel(self):
        """LiDAR 360° - ZOOMED to 20m"""
        self.ax_lidar.clear()
        self.ax_lidar.set_facecolor('#0a0a0a')
        
        snapshot = get_lidar_360()
        if snapshot is None:
            return []
        
        GROUND_Z = -2.5
        above_ground = snapshot[(snapshot[:, 2] > GROUND_Z + 0.2) & (snapshot[:, 2] < GROUND_Z + 3.5)]
        
        if len(above_ground) > 0:
            heights = above_ground[:, 2] - GROUND_Z
            colors = plt.cm.plasma(np.clip(heights / 3.0, 0, 1))
            self.ax_lidar.scatter(above_ground[:, 1], above_ground[:, 0],
                                 c=colors, s=2, alpha=0.5, edgecolors='none')
        
        # Red arrow ego
        arrow = mpatches.Polygon([[0, 2.2], [-1.5, -2], [0, -1.5], [1.5, -2]],
                                fill=True, facecolor='#ff3333', edgecolor='#ff0000', linewidth=2)
        self.ax_lidar.add_patch(arrow)
        
        detected = detect_vehicles_from_lidar(snapshot, debug=False)
        
        # Draw vehicles
        for i, v in enumerate(detected, 1):
            display_x = max(v['size_x'], 3.0)
            display_y = max(v['size_y'], 1.5)
            
            distance_ratio = min(v['distance'] / 30.0, 1.0)
            box_color = plt.cm.RdYlGn_r(distance_ratio)
            
            rect = mpatches.Rectangle((v['y'] - display_y/2, v['x'] - display_x/2),
                                     display_y, display_x, fill=False, 
                                     edgecolor=box_color, linewidth=2)
            self.ax_lidar.add_patch(rect)
            
            # Labels
            direction = "F" if v['x'] > 0 else "B"
            if abs(v['y']) > 0.5:
                direction += "R" if v['y'] > 0 else "L"
            
            self.ax_lidar.text(v['y'], v['x'] + display_x/2 + 0.5,
                              f"{direction}\n{v['distance']:.0f}m",
                              color='white', fontsize=8, ha='center', weight='bold')
        
        # Distance circles - only 10 and 20m
        for r in [10, 20]:
            circle = plt.Circle((0, 0), r, fill=False, edgecolor='#2a4a6a',
                               linestyle='--', linewidth=1, alpha=0.4)
            self.ax_lidar.add_patch(circle)
            
            # Add distance label
            self.ax_lidar.text(1, r-0.5, f'{r}m', color='#4a9eff', fontsize=8, 
                              weight='bold', alpha=0.8,
                              bbox=dict(boxstyle='round,pad=0.2', facecolor='#0a0a0a', 
                                       edgecolor='#2a4a6a', alpha=0.7))
        
        # ZOOMED to 20m range
        self.ax_lidar.set_xlim(-22, 22)
        self.ax_lidar.set_ylim(-22, 22)
        self.ax_lidar.set_aspect('equal')
        self.ax_lidar.set_title('LiDAR 360° (20m)', color='white', fontsize=12, weight='bold')
        self.ax_lidar.tick_params(colors='white', labelsize=8)
        
        return detected
    
    def update_camera_panel(self):
        """Get camera detections (no display)"""
        if latest_image is None:
            return None
        
        results = self.yolo_model(latest_image)
        detections = parse_detections(results)
        
        return detections
    
    def update_attention_panel(self, camera_detections):
        """Grad-CAM with BLUE colormap"""
        self.ax_attention.clear()
        self.ax_attention.set_facecolor('#0a0a0a')
        
        if latest_image is None or camera_detections is None or len(camera_detections) == 0:
            self.ax_attention.text(0.5, 0.5, 'No Detections', ha='center', va='center',
                                  color='white', fontsize=14, transform=self.ax_attention.transAxes)
            return
        
        attention_map = generate_yolo_attention_map(latest_image, self.yolo_model, camera_detections)
        attention_map_blurred = cv2.GaussianBlur(attention_map, (31, 31), 0)
        
        image_rgb = cv2.cvtColor(latest_image, cv2.COLOR_BGR2RGB)
        image_normalized = image_rgb.astype(np.float32) / 255.0
        
        # BLUE colormap (jet)
        heatmap_colored = plt.cm.jet(attention_map_blurred)[:, :, :3]
        
        overlay = 0.6 * image_normalized + 0.4 * heatmap_colored
        overlay = np.clip(overlay, 0, 1)
        
        self.ax_attention.imshow(overlay)
        self.ax_attention.set_title('Grad-CAM', color='white', fontsize=12, weight='bold')
        self.ax_attention.axis('off')
    
    def update_shap_panel(self, features, shap_values, action):
        """SHAP Feature Importance"""
        self.ax_shap.clear()
        self.ax_shap.set_facecolor('#0a0a0a')
        
        prediction = self.label_encoder.transform([action])[0]
        shap_for_class = shap_values[prediction][0]
        
        # Top 6 features
        indices = np.argsort(np.abs(shap_for_class))[::-1][:6]
        
        top_features = [self.feature_columns[i] for i in indices]
        top_shap = [shap_for_class[i] for i in indices]
        top_vals = [features[feat] for feat in top_features]
        
        colors = ['#ff4444' if s < 0 else '#44ff44' for s in top_shap]
        y_pos = np.arange(len(top_features))
        
        self.ax_shap.barh(y_pos, top_shap, color=colors, alpha=0.8, edgecolor='white', linewidth=1)
        self.ax_shap.set_yticks(y_pos)
        
        # Feature labels
        labels = [f"{feat}\n={val:.1f}" for feat, val in zip(top_features, top_vals)]
        self.ax_shap.set_yticklabels(labels, fontsize=9, color='white')
        
        self.ax_shap.set_xlabel('Impact on Decision', fontsize=10, weight='bold', color='white')
        self.ax_shap.set_title(f'SHAP Explanation: {action.upper()}', color='white', fontsize=12, weight='bold')
        self.ax_shap.axvline(x=0, color='white', linestyle='--', linewidth=1)
        self.ax_shap.tick_params(colors='white', labelsize=8)
        self.ax_shap.grid(True, alpha=0.2, color='gray')
    
    def update_decision_panel(self, action, probabilities, features, lidar_detections, camera_detections):
        """Decision + Info COMBINED"""
        self.ax_decision.clear()
        self.ax_decision.set_facecolor('#0a0a0a')
        
        action_colors = {
            'brake': '#ff4444',
            'accelerate': '#44ff44',
            'maintain': '#ffaa00'
        }
        color = action_colors.get(action, '#ffffff')
        
        # BIG ACTION at top
        self.ax_decision.text(0.5, 0.85, action.upper(), 
                             ha='center', va='center', fontsize=32, weight='bold',
                             color=color, transform=self.ax_decision.transAxes)
        
        # Confidence
        max_prob = max(probabilities)
        self.ax_decision.text(0.5, 0.70, f'Confidence: {max_prob:.1%}',
                             ha='center', va='center', fontsize=14, color='white',
                             transform=self.ax_decision.transAxes)
        
        # Separator line
        self.ax_decision.plot([0.1, 0.9], [0.62, 0.62], color='white', linewidth=1, 
                             transform=self.ax_decision.transAxes)
        
        # INFO SECTION
        info_text = "SITUATION\n"
        info_text += "=" * 30 + "\n\n"
        info_text += f"Speed: {features['speed']:.1f} km/h\n"
        info_text += f"Closest Front: {features['closest_front']:.1f}m\n"
        info_text += f"Closest Back: {features['closest_back']:.1f}m\n"
        info_text += f"Safe Distance: {'YES' if features['safe_distance_front'] else 'NO'}\n\n"
        info_text += f"Detections:\n"
        info_text += f"  LiDAR: {len(lidar_detections)} vehicles\n"
        info_text += f"  Camera: {len(camera_detections) if camera_detections else 0} objects\n\n"
        
        # Why decision
        info_text += "Reasoning:\n"
        if action == 'brake':
            if features['closest_front'] < 15:
                info_text += f"• Close vehicle ({features['closest_front']:.1f}m)\n"
            if features['speed'] > 30:
                info_text += f"• High speed ({features['speed']:.1f} km/h)\n"
        elif action == 'accelerate':
            if features['closest_front'] > 20:
                info_text += f"• Clear ahead ({features['closest_front']:.1f}m)\n"
            if features['speed'] < 30:
                info_text += f"• Low speed ({features['speed']:.1f} km/h)\n"
        else:
            info_text += "• Maintaining current state\n"
        
        info_text += f"\n{datetime.now().strftime('%H:%M:%S')} | Frame {self.frame_count}"
        
        # Display info
        self.ax_decision.text(0.05, 0.55, info_text, 
                            transform=self.ax_decision.transAxes,
                            fontsize=9, verticalalignment='top', 
                            family='monospace', color='white',
                            bbox=dict(boxstyle='round', facecolor='#1a1a1a', 
                                    edgecolor='#00ff00', linewidth=2, alpha=0.9))
        
        # Action history timeline at bottom
        self.action_history.append(action)
        
        if len(self.action_history) > 1:
            history_y = 0.05
            history_x = np.linspace(0.1, 0.9, len(self.action_history))
            
            for x, hist_action in zip(history_x, self.action_history):
                hist_color = action_colors.get(hist_action, '#ffffff')
                self.ax_decision.plot(x, history_y, 'o', color=hist_color, markersize=8,
                                     transform=self.ax_decision.transAxes)
        
        self.ax_decision.axis('off')
    
    def update(self):
        """Update entire dashboard"""
        self.frame_count += 1
        
        try:
            # Get data
            lidar_detections = self.update_lidar_panel()
            camera_detections = self.update_camera_panel()
            self.update_attention_panel(camera_detections)
            
            # Extract features
            features = extract_features()
            X = pd.DataFrame([features])[self.feature_columns]
            
            # Predict action
            prediction = self.shap_model.predict(X)[0]
            probabilities = self.shap_model.predict_proba(X)[0]
            action = self.label_encoder.inverse_transform([prediction])[0]
            
            # SHAP
            shap_values = self.shap_explainer.shap_values(X)
            
            # Update panels
            self.update_shap_panel(features, shap_values, action)
            self.update_decision_panel(action, probabilities, features, lidar_detections, camera_detections)
            
            # Title
            self.fig.suptitle('Explainable AI Dashboard - Autonomous Driving', 
                            color='white', fontsize=14, weight='bold')
            
            plt.draw()
            plt.pause(0.001)
            
            return True
            
        except Exception as e:
            print(f"Error: {e}")
            return False


print("✓ 4-Panel Dashboard with ZOOMED LiDAR (20m range)!")
```
